```python
# Points class

class Point :
    """
    Represents a point in a 2D space
    
    attributes: x, y, name
    
    """
    
    # constructor function
    # The double underscores indicate a private method or variable
    # not to be accessed outside the class (in principle)
    def __init__(self, x=0.0, y=0.0, name = ""):
        self.x = x
        self.y = y
        self.name = name
        self.__private_variable = 42
        
    def __add__(self, other) :
        new_point = Point(self.x+other.x, self.y+other.y)
        return new_point
    
    def print_point(self) :
        print("Point ", self.name, "is", self.x, self.y)
        
    # Note that we don't use self here so don't need to pass it in
    # (This is a static function - it does not require an instance of the class)
    def calculate_distance_between_two_points(A, B) :
        return np.sqrt((A.x - B.x)**2.0 + (A.y - B.y)**2.0)
    
    def move_point(self,dx,dy) :
        self.x += dx
        self.y += dy
        
    def plot_point(self, ax) :
        ax.plot(self.x, self.y, 'o', label=self.name)
        
    def update_name(self, new_name) :
        self.name = new_name
        
        
# Some examples of using the Points class
first_point = Point(1.0,2.0,"P")
first_point.print_point()

second_point = Point(3.0,5.5,"Q")
second_point.move_point(1.0, 2.0)
second_point.print_point()

third_point = first_point + second_point
third_point.update_name("R")
third_point.print_point()

# Use the static function
distance = Point.calculate_distance_between_two_points(first_point,second_point)
print("Distance is ", distance)

plt.plot()
plt.grid()
plt.xlabel("x position")
plt.ylabel("y position")
ax = plt.gcf().gca()
first_point.plot_point(ax)
second_point.plot_point(ax)
third_point.plot_point(ax)
plt.legend()

# Try to access the private variable, even though you shouldn't
#print(first_point.__private_variable) # This won't work - try uncommenting it
print(first_point._Point__private_variable) # This will

# Can get info on the new class - uncomment to see this
# Point?
```

```python
# Rectangle class

# UPDATE ME!

class Rectangle :
    """
    Represents a rectangle in a 2D space
    
    attributes: bottom_left_coord, top_right_coord, bottom_right_coord, top_left_coord
    
    """
    def __init__(self, bottom_left_coord, top_right_coord):
        assert bottom_left_coord[0]<top_right_coord[0], "Point not possible: coordinates are not assigned correctly in the x coordinates!"
        assert bottom_left_coord[1]<top_right_coord[1], "Point not possible: coordinates are not assigned correctly in the y coordinates!"
        self.bottom_left_coord = bottom_left_coord
        self.top_right_coord = top_right_coord
        self.bottom_right_coord = np.array([top_right_coord[0], bottom_left_coord[1]])
        self.top_left_coord =  np.array([bottom_left_coord[0],top_right_coord[1]])
    def reflect_in_x_axis(self):
        self.bottom_left_coord[0], self.bottom_left_coord[1] = -self.bottom_left_coord[0], -self.bottom_left_coord[1]
        self.top_right_coord[0], self.top_right_coord[1] = -self.top_right_coord[0], -self.top_right_coord[1]
        self.bottom_right_coord[0], self.bottom_right_coord[1] = -self.bottom_right_coord[0], -self.bottom_right_coord[1]
        self.top_left_coord[0], self.top_left_coord[1] = -self.top_left_coord[0], -self.top_left_coord[1]
        self.bottom_left_coord, self.top_left_coord = self.top_left_coord, self.bottom_left_coord
        self.bottom_right_coord, self.top_right_coord = self.top_right_coord, self.bottom_right_coord
    def reflect_in_y_axis(self):
        self.bottom_left_coord[0]= -self.bottom_left_coord[0]
        self.top_right_coord[0] = -self.top_right_coord[0]
        self.bottom_right_coord[0] = -self.bottom_right_coord[0]
        self.top_left_coord[0] = -self.top_left_coord[0]
        self.top_left_coord, self.top_right_coord = self.top_right_coord, self.top_left_coord
        self.bottom_left_coord, self.bottom_right_coord = self.bottom_right_coord, self.bottom_left_coord
    def print_rectangle(self) :
        print(self.bottom_left_coord[0], self.bottom_left_coord[1])
        print(self.bottom_right_coord[0], self.bottom_right_coord[1])
        print(self.top_left_coord[0], self.top_left_coord[1])
        print(self.top_right_coord[0], self.top_right_coord[1])
    def grow_rectangle(self, magnitude):
        self.top_right_coord = self.bottom_left_coord + magnitude*(self.top_right_coord - self.bottom_left_coord)
        self.bottom_right_coord = self.bottom_left_coord + magnitude*(self.bottom_right_coord - self.bottom_left_coord)
        self.top_left_coord = self.bottom_left_coord + magnitude*(self.top_left_coord - self.bottom_left_coord)

    def plot_rectangle(self):
        plt.plot(np.array([self.bottom_left_coord[0], self.bottom_right_coord[0]]), np.array([self.bottom_left_coord[1], self.bottom_right_coord[1]]), '-', color = "blue")
        plt.plot(np.array([self.bottom_left_coord[0], self.top_left_coord[0]]), np.array([self.bottom_left_coord[1], self.top_left_coord[1]]), '-', color = "blue")
        plt.plot(np.array([self.top_left_coord[0], self.top_right_coord[0]]), np.array([self.top_left_coord[1], self.top_right_coord[1]]), '-', color = "blue")
        plt.plot(np.array([self.bottom_right_coord[0], self.top_right_coord[0]]), np.array([self.bottom_right_coord[1], self.top_right_coord[1]]), '-', color = "blue")
plt.plot()
plt.grid()
plt.xlabel("x position")
plt.ylabel("y position")


test_rectangle = Rectangle(np.array([1,1]), np.array([4,2]))
test_rectangle.print_rectangle()
test_rectangle.plot_rectangle()
plt.legend()
    
```

### ACTIVITY 3:

We now consider a simple system of coupled ODEs. The Lotka–Volterra equations describe predator–prey interactions between a prey population $x$ (for example, rabbits) and a predator population $y$ (for example, foxes). In the absence of predators, the prey population grows at a rate proportional to its size. In the absence of prey, the predator population declines at a rate proportional to its size. Predation occurs at a rate proportional to the product $xy$, reducing the prey population while supporting predator reproduction. The populations are therefore described by the following first-order, autonomous, nonlinear, coupled system of ordinary differential equations:

$\frac{dx}{dt} = ax - bxy$

$\frac{dy}{dt} = -cy + dxy$

Here, $a$ is the prey growth rate, $b$ measures predation, $c$ is the predator death rate, and $d$ measures how effectively consumed prey contributes to predator growth.

What parts or features of the differential equation tell us that it is:
1. First order
2. Autonomous
3. Non linear
4. Coupled 

?

Answers: ... UPDATE ME!
1. First order: The highest deritive used is first derivative.
2. Autonomous: There is no independent value, t, in any of the coefficients or functions on the right side of the equations.
3. Each equation has an xy term and to be linear there can only be first power terms, that are not multiplied together.
4. They are coupled because the evolution of each variable relies on the other variable, as well as it's own. 

### ACTIVITY 4:

The following code solves the Logistic Equation 

$\frac{dy}{dt} = y (1 - y)$

Using a simple explicit integration and the scipy function solve_ivp() which used Runge-Kutta 4th/5th order time integration. Copy it and update it to solve instead the Lotka-Volterra equations. 

Note the following:

1. Why do I substract delta_t when calculating the vector of t_solution points?
2. the solve_ivp method uses the labels t and y for the independent and dependent variables. The y value is assumed to be a vector, allowing you to solve for multiple dependent variables at once, e.g. you can make it np.array([x,y])

```python
# Solve the 1d logistic equation from class
from scipy.integrate import solve_ivp

# Note that the function has to take t as the first argument and y as the second
def calculate_logistic_dydt(t, y):
    """Returns the gradient dx/dt for the logistic equation"""
    dydt = y*(1 - y)
    return dydt

max_time = 10.0
N_time_steps = 25
delta_t = max_time / N_time_steps
t_solution = np.linspace(0.0, max_time-delta_t, N_time_steps) # values of independent variable 
y0 = np.array([0.5]) # an initial condition, y(0) = y0, note it needs to be an array

solution = solve_ivp(calculate_logistic_dydt, [0,max_time], y0,
                        method='RK45', t_eval=t_solution)
 
plt.grid()
plt.xlabel("t", fontsize=16)
plt.ylabel("y", fontsize=16) 
plt.plot(solution.t, solution.y[0], 'o', label="solve_ivp")

# Now do it the "cheap" way 
# increase the number of steps to see how the solution changes
y_solution = np.zeros_like(t_solution)
y_solution[0] = y0[0]
for itime, time in enumerate(t_solution) :
    if itime > 0 :
        dydt = calculate_logistic_dydt(time, y_solution[itime-1])
        y_solution[itime] = y_solution[itime-1] + dydt * delta_t
    
plt.plot(t_solution, y_solution, '-.',label="cheap method")

# Now plot the true solution
A = 1.0/y0 - 1.0
y_true = 1.0 / (1.0 + A * np.exp(-t_solution))
plt.plot(t_solution, y_true, '-', label="true solution")
plt.legend(loc='best');
```
